```python
import numpy as np
import sys
```

```python
trainX = np.loadtxt("./Fall25-ML-hw3/gisette/gisette_trainSet.txt")
trainY = np.loadtxt("./Fall25-ML-hw3/gisette/gisette_trainLabels.txt")
testX = np.loadtxt("./Fall25-ML-hw3/gisette/gisette_testSet.txt")
testY = np.loadtxt("./Fall25-ML-hw3/gisette/gisette_testLabels.txt")
```

```python
# def EuclideanDistance(a:np.array,b:np.array) -> float:
#   return np.sqrt(np.sum((a-b)**2))

def DistanceFromAll(point:np.array, data:np.array):
  distances = np.sqrt(np.sum((data - point)**2, axis=1))
  return np.argsort(distances)

def Classification(point:np.array, data:np.array, label:np.array, k:int):
  dist =  DistanceFromAll(point, data)[:k]
  kNearest = label[dist]
  
  labels = np.unique(label)
  counter = np.zeros(labels.shape, dtype=[('key',labels.dtype),('val',np.int16)])
  counter['key'] = labels
  for i, label in enumerate(dist):
    counter['val'][i] += np.sum(kNearest == label)
  #print(dist)
  #print(counter)
  return counter['key'][np.argmax(counter['val'])]  

def getError(predict:np.array, predictLabel:np.array, data:np.array, dataLabel:np.array, k:int):
  error = 0
  for i in range(len(predictLabel)):
    prediction = Classification(predict[i], data, dataLabel, k)
    if prediction != predictLabel[i]:
      error += 1
  return (error, error/len(predictLabel))

def train_fit(data, label, k):
  splitX = np.split(data,5,axis=0)
  splitY = np.split(label,5,axis=0)
  
  totalErr = 0
  for i in range(5):
    predict, predictLabel = splitX[i], splitY[i]
    data = np.concatenate([splitX[j] for j in range(5) if j != i])
    dataLabel = np.concatenate([splitY[j] for j in range(5) if j != i])
    totalErr += getError(predict,predictLabel,data,dataLabel,k)[1]
  cvErr = totalErr/5
  return cvErr

def getBestK(data,label):
  error = sys.float_info.max
  k = 0
  for i in range(1,11):
    cvError = train_fit(data,label,i)
    if cvError < error:
      error = cvError
      k = i
    print(f"k={i}     cross valindation error = {cvError}")
  return (k,error)
    
```

```python
getBestK(trainX,trainY)
```
