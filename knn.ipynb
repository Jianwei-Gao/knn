```python
def EuclideanDistance(a:np.array,b:np.array) -> float:
  return np.sqrt(np.sum((a-b)**2))

def DistanceFromAll(point:np.array, data:np.array):
  distances = np.zeros(data.shape[0], dtype=[('i',np.int16),('d',np.float32)])
  distances['i'] = np.arange(1,trainX.shape[0]+1)
  for i, val in enumerate(data):
    distances['d'][i] = EuclideanDistance(point,data[i])
  return np.sort(distances, order='d')

def Classification(point:np.array, data:np.array, label:np.array, k:int):
  dist =  DistanceFromAll(point, data)[0:k]
  counter = np.zeros(label.shape)
  for row in dist:
    print(row)
  return dist  

```
